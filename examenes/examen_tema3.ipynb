```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline

# Asegúrate de encontrarte en el directorio donde se encuentra el archivo. 
os.chdir("/home/gossbosther/ml-course/examenes/data")     # Reemplaza el path con el tuyo.
df = pd.read_csv('data_coffeeshop.csv', header='infer', na_values=[' '])
print(df.shape)
df.head(5)
```

1) Genera una tabla de frecuencias para la variable 'yearOfStart' y visualízala por año.
    • Ordena por año.
    • Dibuja una gráfica de línea de 1997 a 2014. <= Pista: plt.xlim()

```python
tabla_frecuencias = df["Year of Start"].value_counts().sort_index()
tabla_frecuencias.plot(kind='line', title='Frecuencia por Anio de Inicio')
plt.xlim(1997, 2014)
plt.show()
```

2) Divide los datos por el estado actual del negocio ('In' o 'Out'). Después, visualiza la tendencia anual de las frecuencias de la variable 'yearOfStart'.
    • Ordena por año.
    • Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.
    • Usa el objeto 'figure'.

```python
df_in = df[df['Current State'] == 'In']
df_out = df[df['Current State'] == 'Out']
frec_in = df_in['Year of Start'].value_counts().sort_index()
frec_out = df_out['Year of Start'].value_counts().sort_index()

fig = plt.figure(figsize=(10, 5))
plt.plot(frec_in.index, frec_in.values, label='In')
plt.plot(frec_out.index, frec_out.values, label='Out')
plt.xlim(1997, 2014)
plt.legend(title="Current State")
plt.title("Tendencia anual por estado del negocio")
plt.show()
```
