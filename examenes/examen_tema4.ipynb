#### 2. Probabilidad binomial
Una familia tiene tres hijos. Suponiendo que cada hijo tiene un 50% de probabilidad de ser varón, independientemente de los demás, responde:

a) ¿Cuál es la probabilidad de tener exactamente dos hijos varones?

Para calcular esto, utilizamos la fórmula de la probabilidad binomial:

$$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$

Donde $k = 2$ (dos hijos varones):

$$P(X = 2) = \binom{3}{2} (0.5)^2 (0.5)^{3-2}$$
$$P(X = 2) = \frac{3!}{2!(3-2)!} (0.25) (0.5)^1$$
$$P(X = 2) = 3 \cdot 0.25 \cdot 0.5$$
$$P(X = 2) = 0.375$$

La probabilidad de tener exactamente dos hijos varones es del **37.5%**.


b) ¿Cuál es el número esperado (media) de hijos varones?

La fórmula para la media o valor esperado en una distribución binomial es:

$$\mu = n \cdot p$$
$$\mu = 3 \cdot 0.5$$
$$\mu = 1.5$$

El número esperado de hijos varones es **1.5**.

c) ¿Cuál es la varianza del número de hijos varones?

La fórmula para la varianza en una distribución binomial es:

$$\sigma^2 = n \cdot p \cdot (1-p)$$
$$\sigma^2 = 3 \cdot 0.5 \cdot 0.5$$
$$\sigma^2 = 1.5 \cdot 0.5$$
$$\sigma^2 = 0.75$$

La varianza del número de hijos varones es **0.75**.

#### 5. Intervalo de confianza
Se seleccionan aleatoriamente 30 estudiantes de una institución educativa y se registran sus calificaciones. Los resultados son:
- Tamaño de la muestra: n = 30
- Media muestral: 80
- Varianza muestral: 9
- Nivel de confianza: 95%

Calcula un intervalo de confianza del 95% para la media poblacional de las calificaciones. Muestra el procedimiento e interpreta el resultado.

*   Tamaño de la muestra ($n$): $30$
*   Media muestral ($\bar{x}$): $80$
*   Varianza muestral ($\sigma^2$): $9$ (Por lo tanto, la desviación estándar es $\sigma = \sqrt{9} = 3$)
*   Nivel de confianza: $95\%$ (El valor crítico correspondiente es $Z = 1.96$)

Utilizamos la formula para el intervalo de confianza de la media:
$$IC_{95\%} = \bar{x} \pm Z \frac{\sigma}{\sqrt{n}}$$

Sustituyendo los valores en la fórmula:
$$IC_{95\%} = 80 \pm 1.96 \frac{3}{\sqrt{30}}$$

Calculamos el error estándar:
$$IC_{95\%} = 80 \pm 1.96 \left(\frac{3}{5.4772}\right)$$
$$IC_{95\%} = 80 \pm 1.96 (0.5477)$$

Calculamos el margen de error:
$$IC_{95\%} = 80 \pm 1.0735$$

Finalmente, obtenemos los límites del intervalo:
* Límite inferior: $80 - 1.0735 = 78.9265$
* Límite superior: $80 + 1.0735 = 81.0735$

El intervalo de confianza es $(78.93, 81.07)$.

Con un nivel de confianza del $95\%$, se estima que la verdadera media poblacional de las calificaciones de todos los estudiantes de la institución se encuentra entre **$78.93$** y **$81.07$**.


#### 6. Correlación de variables utilizando Python
Utiliza el conjunto de datos load_breast_cancer de Scikit-learn para identificar qué característica tiene la correlación lineal más fuerte (positiva o negativa) con la variable objetivo target.
Código de partida:


```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Cargar los datos
data = load_breast_cancer()
# Crear un DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)
# Agregar la variable objetivo
df["target"] = 1 - data.target
```

Actividades: 

a) Calcula la correlación de Pearson entre cada característica y la variable target.

```python
correlaciones = df.corr(method='pearson')['target']
print(correlaciones)
```

b) Identifica la característica con mayor correlación en valor absoluto, excluyendo la correlación de target consigo misma.

```python
correlaciones_sin_target = correlaciones.drop('target')
caracteristica_max_corr = correlaciones_sin_target.abs().idxmax()
valor_max_corr = correlaciones_sin_target[caracteristica_max_corr]

print(f"La característica con mayor correlación es: '{caracteristica_max_corr}'")
print(f"Valor de la correlación: {valor_max_corr:.4f}")
```
